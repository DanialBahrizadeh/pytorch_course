```python
class InvalidMatrixShape(Exception):
    pass

def matrix_mul(A: torch.Tensor, B: torch.Tensor) -> torch.Tensor:
    if (A.shape[1] != B.shape[0]):
        raise InvalidMatrixShape()
    n,m = A.shape[0], B.shape[1]
    C = torch.zeros(n,m)
    B_T = B.T
    for i in range(n):
        for j in range(m):
            C[i][j] = torch.dot(A[i] , B_T[j])
    return C

```

```python
A = torch.tensor([
    [6,3,5],
    [2,4,1]
])

B = torch.tensor([
    [1,4],
    [2,5],
    [3,6]
])
```

```python
%%time

A @ B
```

```python
%%time
matrix_mul(A,B)
```
