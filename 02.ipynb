```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
from sklearn.datasets import make_circles

print(torch.__version__)
```

```python

n_samples: int = 1000

X, y = make_circles(
    n_samples=n_samples,
    noise= 0.03,
    random_state=42
)
```

```python

plt.scatter(x=X[:, 0],y=X[:, 1], c=y)
```

```python
df = pd.DataFrame({
    "X1": X[:,0],
    "X2": X[:,1],
    "label": y
})

df.head()
```
